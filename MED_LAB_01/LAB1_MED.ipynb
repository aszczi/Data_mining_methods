Zgrupuj dane według kolumny Node. Policz całkowitą liczbę wystąpień dla każdego węzła. Posortuj wyniki malejąco i wyodrębnij 5 maszyn, które wygenerowały najwięcej zdarzeń w badanym okresie.

3. Identyfikacja stanów systemu - Kolumna State przechowuje informacje o statusie danego zdarzenia w klastrze. Zidentyfikuj wszystkie unikalne wartości występujące w kolumnie State. **Rozwiązanie**

```python
import pandas as pd

# Wczytanie pliku
df = pd.read_csv("HPC_2k.log_structured.csv")

# 1. Konwersja czasu i najaktywniejsza godzina
df['Datetime'] = pd.to_datetime(df['Time'], unit='s')
df['Hour'] = df['Datetime'].dt.hour
hour_counts = df['Hour'].value_counts().sort_index()
max_hour = hour_counts.idxmax()
max_hour_count = hour_counts.max()
print(f"1.")
print(f"Max activity hour: {max_hour} with {max_hour_count} logs")
print(hour_counts)
print('\n')

# 2. 5 najbardziej awaryjnych węzłów
node_counts = df['Node'].value_counts().head(5)
print(f"2.")
print(node_counts)
print('\n')

# 3. Procentowy udział stanów systemu
state_counts = df['State'].value_counts()
total_logs = len(df)
state_percentages = (state_counts / total_logs * 100).round(2)
print(f"3.")
for state, pct in state_percentages.items():
    print(f"{state}: {pct:.2f}%")
```
